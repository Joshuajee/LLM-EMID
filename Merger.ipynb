```python
import pandas as pd
```

```python
# Read the dataset
clean_df = pd.read_csv('./hidden-data-2/Motor Insurance - Dataset(Cleaned_Safety_Prompts)-GEM.csv')
clean_df
```

```python
faulty_df = pd.read_csv("./faulty_subcats.csv")
faulty_df
```

```python
# set index
clean_df.set_index("Index")
faulty_df.set_index("Index")
```

```python
faulty_df = faulty_df[["Index", "Subcat"]]
faulty_df
```

```python
new_df = clean_df.merge(faulty_df, on="Index", how="left", )
new_df
```

```python
new_df["Subcat_x"] = new_df.apply(lambda row: row["Subcat_y"] if row["Subcat_x"] == None else row["Subcat_x"], axis=1)
new_df
```

```python
new_df.drop(["Subcat_y"], inplace=True, axis=1)
```

```python
new_df.rename(columns={'Subcat_y': 'Subcat'}, inplace=True)
```

```python
new_df.to_csv('./hidden-data-final/Motor Insurance - Dataset(Cleaned_Safety_Prompts)-GEM.csv')
```
